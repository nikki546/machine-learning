```python
'''1. What is a parameter?
A parameter is a numerical value that represents a characteristic of a population or a model.
In Machine Learning, parameters are learned automatically from training data.
They help the model make predictions.
Parameters directly affect the output of the model.
Examples include weights and bias in linear regression.
Once learned, parameters remain fixed during testing.'''
```

```python
'''2. What is correlation?
Correlation is a statistical measure that shows the relationship between two variables.
It indicates how strongly variables move together.
Correlation values range from –1 to +1.
A value close to +1 indicates strong positive correlation.
A value close to –1 indicates strong negative correlation.
A value near 0 means no relationship.'''
```

```python
'''3. What does negative correlation mean?
Negative correlation means that when one variable increases, the other decreases.
Both variables move in opposite directions.
The correlation coefficient is less than zero.
This relationship is linear but inverse.
Example: As the price increases, demand decreases.
It helps understand inverse relationships in data.'''
```

```python
'''4. Define Machine Learning. What are the main components in ML?
Machine Learning is a branch of Artificial Intelligence that enables systems to learn from data without explicit programming.
It focuses on building models that improve with experience.
The main components include data, features, and algorithms.
A model learns patterns from data.
A loss function measures error.
An optimizer updates parameters to reduce loss.'''
```

```python
'''5. How does loss value help determine whether the model is good or not?
Loss value measures the difference between predicted output and actual output.
It indicates how well the model is performing.
Lower loss means predictions are closer to actual values.
Higher loss means the model is making more errors.
During training, the goal is to minimize loss.
Loss helps compare different models objectively.'''
```

```python
'''6. What are continuous and categorical variables?
Continuous variables are numerical values that can take any value within a range.
Examples include height, weight, and temperature.
Categorical variables represent distinct categories or labels.
Examples include gender, color, and city names.
Continuous data is used for regression problems.
Categorical data is common in classification problems.'''
```

```python
'''7. How do we handle categorical variables in Machine Learning?
Categorical variables must be converted into numbers.
Label Encoding assigns a unique number to each category.
One-Hot Encoding creates binary columns for each category.
Ordinal Encoding is used when categories have order.
Encoding allows algorithms to process categorical data.
Proper encoding improves model accuracy.'''
```

```python
'''8. What do you mean by training and testing a dataset?
Training data is used to teach the model patterns.
The model learns parameters from training data.
Testing data is used to evaluate model performance.
Testing data is not seen by the model during training.
This ensures unbiased evaluation.
It helps detect overfitting and underfitting.'''
```

```python
'''9. What is sklearn.preprocessing?
sklearn.preprocessing is a module in Scikit-learn.
It provides tools for data transformation and scaling.
It includes StandardScaler, MinMaxScaler, and LabelEncoder.
It helps normalize numerical data.
It also supports encoding categorical variables.
Preprocessing improves model performance.'''
```

```python
'''10. How do we split data for model fitting in Python?
Data splitting is done using train_test_split.
It divides data into training and testing sets.
It is available in sklearn.model_selection.
Usually 70–80% data is used for training.
Remaining data is used for testing.
Random splitting avoids bias.'''
```

```python
'''11. How do you approach a Machine Learning problem?
First, understand the problem and define objectives.
Collect and clean the data.
Perform Exploratory Data Analysis (EDA).
Select appropriate features.
Choose a suitable ML algorithm.
Train, test, and evaluate the model.'''
```

```python
'''12. Why do we have to perform EDA before fitting a model?
EDA helps understand data structure and patterns.
It identifies missing values and outliers.
It shows relationships between variables.
EDA helps in feature selection.
It improves data quality.
Better EDA leads to better models.'''
```

```python
'''13. What is correlation?
Correlation measures the degree of relationship between variables.
It shows whether variables move together or oppositely.
Correlation coefficient ranges from –1 to +1.
Positive correlation means same direction movement.
Negative correlation means opposite direction movement.
Zero correlation means no relationship.'''
```

```python
'''14. What does negative correlation mean?
Negative correlation shows inverse relationship.
As one variable increases, the other decreases.
The correlation coefficient is less than zero.
It indicates opposite trends.
Example: More study time, less failure rate.
It helps identify inverse patterns.'''
```

```python
'''15. How can you find correlation between variables in Python?
Correlation can be found using corr() function.
It is available in pandas DataFrame.
Pearson correlation is commonly used.
It returns correlation coefficient values.
Heatmaps visualize correlation.
It helps in feature selection.'''
```

```python
'''16. What is causation? Explain difference between correlation and causation with example.
Causation means one variable directly affects another.
Correlation only shows relationship, not cause.
Correlation does not imply causation.
Example: Ice cream sales and drowning cases are correlated.
But ice cream does not cause drowning.
Causation requires evidence of direct effect.'''
```

```python
'''17. What is an Optimizer? Explain types with example.
An optimizer updates model parameters to minimize loss.
It improves learning efficiency.
Common optimizers include Gradient Descent, SGD, Adam.
Gradient Descent updates parameters gradually.
Adam adapts learning rate automatically.
Optimizers help reach optimal solution faster.'''
```

```python
'''18. What is sklearn.linear_model?
sklearn.linear_model is a module for linear models.
It includes Linear Regression and Logistic Regression.
Used for regression and classification tasks.
It is easy to implement.
Supports regularization techniques.
Widely used in ML applications.'''
```

```python
'''19. What does model.fit() do? What arguments are required?
model.fit() trains the model.
It learns parameters from training data.
Required arguments are input features (X) and target (y).
It applies learning algorithm.
Training happens only once.
After fitting, the model is ready to predict.'''
```

```python
'''20. What does model.predict() do? What arguments are required?
model.predict() generates predictions.
It uses trained model parameters.
Input argument is test data or new data.
It returns predicted output.
Used after model fitting.
Helps evaluate model accuracy.'''
```

```python
'''22. What is feature scaling? How does it help in ML?
Feature scaling standardizes numerical data.
It brings all features to same scale.
Prevents dominance of large values.
Improves algorithm performance.
Important for distance-based algorithms.
Examples include normalization and standardization.'''
```

```python
'''23. How do we perform scaling in Python?
Scaling is done using StandardScaler or MinMaxScaler.
They are available in sklearn.preprocessing.
Fit scaler on training data.
Transform both train and test data.
Prevents data leakage.
Improves model accuracy.'''
```

```python
'''24. How do we split data for model fitting in Python?
Data is split using train_test_split.
It is imported from sklearn.model_selection.
Splits features and labels.
Ensures random distribution.
Helps test generalization.
Common split ratio is 80:20.'''
```
